```python

testfile = 'test_list'
with open(testfile, 'r') as otf:
    lines = otf.readlines()
testlist = []
for line in lines:
    testlist.append(line.strip())
```

```python
normoutlist = []
for idx in range(len(testlist)):
    text = testlist[idx]
    #print(text)
    normtext = normalizer.normalize(text, lang='mix')
    normoutlist.append(normtext)
```

```python

outfile = 'outfile'
oof = open(outfile, 'w')
for idx in range(len(normoutlist)):
    normtext = normoutlist[idx]
    oof.write(normtext + '\n')
oof.close()
```
